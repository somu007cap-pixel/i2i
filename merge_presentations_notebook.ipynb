# Merge and polish PowerPoint presentations

This notebook merges the content from the source presentation into the draft presentation template and prepares a visually polished deck for an MTech final review.

```python
from pathlib import Path
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN
from pptx.dml.color import RGBColor
import os

print('Libraries imported successfully')
```

```python
base = Path(r'c:\I2I\Bits_Submission\generated_final')
source_path = base / '2024AA05661_Final_Presentation_READY.pptx'
template_path = base / 'Sumit_finalSem_Presentation_draft.pptx'
out_path = base / 'Sumit_finalSem_Presentation_professional.pptx'

source = Presentation(source_path)
template = Presentation(template_path)

print('Loaded source slides:', len(source.slides))
print('Loaded template slides:', len(template.slides))
```

```python
def extract_text(slide):
    texts = []
    for shape in slide.shapes:
        if hasattr(shape, 'text') and shape.text.strip():
            texts.append(' '.join(shape.text.split()))
    return texts

source_texts = [extract_text(slide) for slide in source.slides]
for idx, texts in enumerate(source_texts, 1):
    print(f'Slide {idx}:')
    for text in texts[:8]:
        print('-', text[:220])
    print()
```

```python
def fill_textbox(shape, text, font_size=20, bold=False, color=RGBColor(0x1F, 0x2F, 0x3F), align=PP_ALIGN.LEFT):
    shape.text = ''
    tf = shape.text_frame
    tf.clear()
    p = tf.paragraphs[0]
    p.alignment = align
    run = p.add_run()
    run.text = text
    run.font.size = Pt(font_size)
    run.font.bold = bold
    run.font.color.rgb = color
    run.font.name = 'Arial'


def ensure_title(slide, title):
    for shape in slide.shapes:
        if hasattr(shape, 'text') and 'title' in shape.text.lower() or shape.is_placeholder and shape.placeholder_format.idx == 0:
            pass


def style_slide(slide):
    for shape in slide.shapes:
        if hasattr(shape, 'text') and shape.text.strip():
            if shape.text_frame.paragraphs:
                for p in shape.text_frame.paragraphs:
                    for run in p.runs:
                        run.font.name = 'Arial'
                        run.font.size = Pt(20)
                        run.font.color.rgb = RGBColor(0x1F, 0x2F, 0x3F)

print('Helpers ready')
```

```python
# Map the source slide content into the template structure.
# For a polished review deck, we preserve the template layout and replace the body text with the strongest content from the source.

max_slides = min(len(source.slides), len(template.slides))
for i in range(max_slides):
    src_slide = source.slides[i]
    dst_slide = template.slides[i]
    src_texts = extract_text(src_slide)
    if not src_texts:
        continue

    # Replace text in existing placeholders if present.
    for shape in dst_slide.shapes:
        if hasattr(shape, 'text') and shape.text.strip():
            if 'title' in shape.text.lower() and src_texts:
                fill_textbox(shape, src_texts[0][:120], font_size=28, bold=True, color=RGBColor(0x10, 0x2B, 0x5E))
            elif len(src_texts) > 1:
                fill_textbox(shape, '\n'.join(src_texts[1:4]), font_size=18, color=RGBColor(0x1F, 0x2F, 0x3F))
                break
    style_slide(dst_slide)

print('Mapped content into template slides')
```

```python
# Refine slide titles and body text for a professional academic review deck.
for i, slide in enumerate(template.slides):
    for shape in slide.shapes:
        if hasattr(shape, 'text') and shape.text.strip():
            text = ' '.join(shape.text.split())
            if text.lower().startswith('title'):
                continue
            if len(text) > 220:
                # Keep it concise and presentation-friendly.
                shape.text = text[:220] + '...'

print('Refined slide text for readability')
```

```python
template.save(out_path)
print('Saved professional presentation to:', out_path)
```

```python
from pathlib import Path
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.enum.text import PP_ALIGN
from pptx.enum.shapes import MSO_AUTO_SHAPE_TYPE
from pptx.dml.color import RGBColor

out_path = Path(r'c:\I2I\Bits_Submission\generated_final\Sumit_finalSem_Presentation_clean.pptx')
prs = Presentation()
prs.slide_width = Inches(13.333)
prs.slide_height = Inches(7.5)

NAVY = RGBColor(31, 78, 121)
BLUE = RGBColor(47, 117, 181)
DARK = RGBColor(38, 38, 38)
GRAY = RGBColor(89, 89, 89)
WHITE = RGBColor(255, 255, 255)
LIGHT = RGBColor(244, 247, 251)


def add_bg(slide, color=WHITE):
    shape = slide.shapes.add_shape(MSO_AUTO_SHAPE_TYPE.RECTANGLE, 0, 0, prs.slide_width, prs.slide_height)
    shape.fill.solid()
    shape.fill.fore_color.rgb = color
    shape.line.color.rgb = color


def add_header(slide, title, subtitle=None, slide_no=None):
    add_bg(slide)
    bar = slide.shapes.add_shape(MSO_AUTO_SHAPE_TYPE.RECTANGLE, 0, 0, prs.slide_width, Inches(0.58))
    bar.fill.solid()
    bar.fill.fore_color.rgb = NAVY
    bar.line.color.rgb = NAVY

    title_box = slide.shapes.add_textbox(Inches(0.65), Inches(0.8), Inches(10.8), Inches(0.65))
    tf = title_box.text_frame
    tf.clear()
    p = tf.paragraphs[0]
    p.alignment = PP_ALIGN.LEFT
    run = p.add_run()
    run.text = title
    run.font.name = 'Arial'
    run.font.size = Pt(22)
    run.font.bold = True
    run.font.color.rgb = NAVY

    if subtitle:
        sub_box = slide.shapes.add_textbox(Inches(0.65), Inches(1.38), Inches(10.8), Inches(0.35))
        tf2 = sub_box.text_frame
        tf2.clear()
        p2 = tf2.paragraphs[0]
        p2.alignment = PP_ALIGN.LEFT
        run2 = p2.add_run()
        run2.text = subtitle
        run2.font.name = 'Arial'
        run2.font.size = Pt(11)
        run2.font.color.rgb = GRAY

    if slide_no is not None:
        num_box = slide.shapes.add_textbox(Inches(12.0), Inches(0.15), Inches(0.9), Inches(0.25))
        tf3 = num_box.text_frame
        tf3.clear()
        p3 = tf3.paragraphs[0]
        run3 = p3.add_run()
        run3.text = str(slide_no)
        run3.font.name = 'Arial'
        run3.font.size = Pt(10)
        run3.font.bold = True
        run3.font.color.rgb = WHITE


def add_metric_card(slide, title, value, x, y, w=2.0, h=1.3):
    card = slide.shapes.add_shape(MSO_AUTO_SHAPE_TYPE.ROUNDED_RECTANGLE, Inches(x), Inches(y), Inches(w), Inches(h))
    card.fill.solid()
    card.fill.fore_color.rgb = LIGHT
    card.line.color.rgb = BLUE

    tb = slide.shapes.add_textbox(Inches(x + 0.12), Inches(y + 0.12), Inches(w - 0.24), Inches(0.3))
    tf = tb.text_frame
    tf.clear()
    p = tf.paragraphs[0]
    p.alignment = PP_ALIGN.CENTER
    run = p.add_run()
    run.text = title
    run.font.name = 'Arial'
    run.font.size = Pt(10)
    run.font.bold = True
    run.font.color.rgb = GRAY

    tb2 = slide.shapes.add_textbox(Inches(x + 0.12), Inches(y + 0.45), Inches(w - 0.24), Inches(0.45))
    tf2 = tb2.text_frame
    tf2.clear()
    p2 = tf2.paragraphs[0]
    p2.alignment = PP_ALIGN.CENTER
    run2 = p2.add_run()
    run2.text = value
    run2.font.name = 'Arial'
    run2.font.size = Pt(16)
    run2.font.bold = True
    run2.font.color.rgb = NAVY


def add_bullets(slide, bullets, x=0.85, y=1.85, w=5.9, h=4.8):
    box = slide.shapes.add_textbox(Inches(x), Inches(y), Inches(w), Inches(h))
    tf = box.text_frame
    tf.clear()
    tf.word_wrap = True
    for i, bullet in enumerate(bullets):
        p = tf.paragraphs[0] if i == 0 else tf.add_paragraph()
        p.text = bullet
        p.level = 0
        p.font.name = 'Arial'
        p.font.size = Pt(18)
        p.font.color.rgb = DARK
        p.space_after = Pt(8)
        p.bullet = True


def add_image(slide, path, x, y, w, h):
    if Path(path).exists():
        slide.shapes.add_picture(str(path), Inches(x), Inches(y), width=Inches(w), height=Inches(h))


def add_footer(slide, text):
    box = slide.shapes.add_textbox(Inches(0.65), Inches(7.0), Inches(10.0), Inches(0.25))
    tf = box.text_frame
    tf.clear()
    p = tf.paragraphs[0]
    run = p.add_run()
    run.text = text
    run.font.name = 'Arial'
    run.font.size = Pt(9)
    run.font.color.rgb = GRAY


def add_slide(title, subtitle, bullets, slide_no, footer, image=None, metrics=None):
    slide = prs.slides.add_slide(prs.slide_layouts[6])
    add_header(slide, title, subtitle, slide_no)
    if image:
        add_image(slide, image, 7.0, 1.15, 5.7, 3.8)
    if metrics:
        for label, value, x, y in metrics:
            add_metric_card(slide, label, value, x, y)
    add_bullets(slide, bullets, x=0.85, y=5.15 if image else 1.85, w=6.0 if image else 6.0)
    add_footer(slide, footer)
    return slide

base = Path(r'c:\I2I\Bits_Submission\generated_final')
diag = base / 'diagrams'
fig = Path(r'c:\I2I\seizure_detection\outputs\dissertation_figures')

slides = [
    ("Tonic-Clonic Seizure Event Detection from Wearable Physiological Time-Series Data", "Final Dissertation Presentation", [
        "Sumit Kumar | 2024AA05661",
        "Infineon Technologies, Bengaluru",
        "BITS Pilani WILP | M.Tech. AIML | July 2026",
    ], 1, "2024AA05661 | Final Dissertation Viva", None, None),
    ("Problem Context and Scope", "Why this study matters", [
        "Goal: detect tonic-clonic seizure events from wearable physiological time-series data.",
        "Signals used: ACC, BVP, HR, EDA, and TEMP from Empatica recordings.",
        "Scope: offline patient-held-out prototype evaluation.",
        "This work is a feasibility study, not a clinical validation study.",
    ], 2, "Problem definition", diag / '04_evaluation_flow.png', None),
    ("Research Objectives", "What the dissertation sets out to achieve", [
        "Develop a reproducible preprocessing pipeline for multi-rate wearable signals.",
        "Construct fixed-length windows using transparent weak-label assumptions.",
        "Design and evaluate Standard and Pro TSMixer variants.",
        "Assess performance using event-level and alert-level metrics.",
    ], 3, "Objectives", diag / '01_project_phase_flow.png', None),
    ("Why Accuracy Is Not Enough", "Rare-event evaluation requires better metrics", [
        "Seizure windows are extremely rare, making the task highly imbalanced.",
        "A normal-only classifier can appear highly accurate while missing all events.",
        "The all-normal baseline reached 0.9988 accuracy, which is misleading here.",
        "Interpretation relies on AUC, recall, and false alarms per hour.",
    ], 4, "Evaluation insight", fig / '01_class_imbalance_log_scale.png', [("All-normal baseline", "0.9988", 7.25, 1.3), ("Held-out events", "6", 9.8, 1.3)]),
    ("Dataset and Label Policy", "Data and annotation design", [
        "Patient-held-out split was used to reduce leakage risk.",
        "Seizure labels provide onset information but not reliable offsets.",
        "Onset-derived weak labels were used and explicitly disclosed.",
        "The evaluation protocol is transparent and reproducible.",
    ], 5, "Data and annotation design", diag / '01_project_phase_flow.png', None),
    ("Proposed Model Design", "Standard and Pro TSMixer tiers", [
        "The Standard tier acts as the accessible baseline.",
        "The Pro tier uses additional physiological channels.",
        "The Pro tier is not assumed better by design.",
        "It is retained only when validation evidence supports improvement.",
    ], 6, "Model design", diag / '02_tsmixer_standard_pro_io.png', None),
    ("Experimental Protocol", "A transparent and reproducible workflow", [
        "Training used observed windows from the training split only.",
        "Thresholds and promoted variants were selected on validation data.",
        "Alert episodes used refractory post-processing to reduce repeated alarms.",
        "The final claim is bounded to this dataset and weak-label protocol.",
    ], 7, "Protocol", fig / '04_alarm_budget_sweep.png', None),
    ("Key Results", "Main empirical findings", [
        "Standard AUC: 0.4259",
        "Pro AUC: 0.5485",
        "Standard recall: 0.0276",
        "Pro recall: 0.1614",
        "Pro detected 3/6 events compared with 1/6 for Standard.",
    ], 8, "Results", fig / '02_standard_vs_pro_metrics.png', [("Standard AUC", "0.4259", 7.25, 1.3), ("Pro AUC", "0.5485", 9.8, 1.3)]),
    ("Baseline Comparison", "Positioning the proposed approach", [
        "CNN-LSTM was the strongest empirical detector among the baselines.",
        "TSMixer Pro was retained as the compact edge-oriented candidate.",
        "The comparison helped separate the best detector from the proposed lightweight architecture.",
        "The findings support refinement rather than overclaiming deployment readiness.",
    ], 9, "Benchmarking", fig / '05_baseline_comparison.png', None),
    ("Robustness and Limitations", "What the evidence supports", [
        "Robustness runs showed sensitivity to weak event-duration assumptions.",
        "Patient-held-out evaluation replaced easier but less reliable splits.",
        "The current evidence is software-level feasibility, not hardware validation.",
        "The study is limited by offline recordings and a small held-out event set.",
    ], 10, "Limitations", fig / '06_ablation_summary.png', None),
    ("Edge Feasibility", "From software prototype to implementation pathway", [
        "The current evidence is software-level feasibility, not hardware validation.",
        "Board latency, memory, and power remain future measurements.",
        "The implementation pathway is defined through INT8 and firmware-facing artifacts.",
        "The next step is measured embedded evaluation on the target device.",
    ], 11, "Feasibility", fig / '07_edge_feasibility.png', None),
    ("Conclusion and Contributions", "A rigorous prototype and feasibility study", [
        "Built an end-to-end, reproducible wearable seizure-event detection pipeline.",
        "The final protocol uses patient-held-out evaluation, weak-label disclosure, and event/alert metrics.",
        "TSMixer Pro improved Standard TSMixer from 1/6 to 3/6 detected events and from AUC 0.4259 to 0.5485.",
        "The work should be positioned as a rigorous prototype and feasibility study, not clinical validation.",
    ], 12, "Conclusion", fig / '08_event_failure_analysis.png', None),
    ("Thank You", "Questions and discussion", [
        "Thank you for your attention.",
        "Questions and discussion are welcome.",
    ], 13, "Q&A", None, None),
]

for title, subtitle, bullets, slide_no, footer, image, metrics in slides:
    add_slide(title, subtitle, bullets, slide_no, footer, image=image, metrics=metrics)

prs.save(out_path)
print('Built presentation using the actual project figures:', out_path)
```
